```python
#Analisis de variables numéricas
for v in variables_numericas:
   sns.histplot(df[v], kde=True, bins=20)
   plt.title("Histograma de {v}")
   plt.show()
```

```python
#Analisis de variables categóricas
for v in variables_categoricas:
    print(f"frecuencia de {v}")
    print(df[v].value_counts())

    sns.countplot(x=v,data=df)
    plt.title(f"Distribución de {v}")
    plt.show()
```

```python
#Correlacion
print("Matriz de Correlación")
df[['Edad', 'Ingresos', 'Casado', 'Satisfaccion', 'NumeroHijos']].corr()
```

```python
sns.heatmap(df[['Edad', 'Ingresos', 'Casado', 'Satisfaccion', 'NumeroHijos']].corr(), annot = True)
plt.title("Mapa de color de Correlación")
plt.show()
```

```python
#5 RElacion entre variables
for v in variables_categoricas:
    sns.scatterplot(x="Edad",y="Ingresos",hue=v, data = df)
    plt.title(f"Relacion entre edad e ingresos por {v}")
    plt.show()
```

```python
#Relacion entre variables categoricas y edad
#la raya en los cuadros es el dato que ma se repite
for v in variables_categoricas:
    sns.boxplot(x=v, y="Edad", data = df)
    plt.title(f"Edad por {v}")
    plt.show()
```

```python
#6. analisis de variables categoricos ordinales
#SELECT NivelEducacion, AVG(Ingresos) FROM df GROUP BY NivelEducacion
print(df.groupby('NivelEducacion')['Ingresos'].mean()  )
```

```python
#6. analisis de variables categoricos ordinales
#linea es el rango de valores que puede haber
for v in variables_categoricas:

    print(df.groupby(v)['Ingresos'].mean())
    sns.barplot(x = v, y = "Ingresos", data = df)
    plt.title(f"Ingresos promedio por {v}")
    plt.xlabel(v)
    plt.ylabel("Ingresos")
    plt.show()

```

```python
#7. Atipicos (outliers)
#para que sea atipico debe ser 1.5 veces mas grande que el bigote derecho
#las rayas como bigotes son el rango cuartil, es el rango, la distribucion de datos, promedio cuartil toma la mediana, que es la rayita en medio del cuadrado
for v in variables_numericas:
  sns.boxplot(x = df[v])
  plt.title(f"Boxplot de {v}")
  plt.show()
```
